```python
import requests
from dotenv import load_dotenv
import os
```

# Criando .csv

## Chave API

```python
load_dotenv()
ELETRICITY_API = os.getenv('ELETRICITY_API')
```

## Requests

```python
url = "https://api.electricitymaps.com/v4/carbon-free-energy/past-range?start=2024-05-19T03%3A00%3A00Z&end=2024-05-19T08%3A00%3A00Z"

response = requests.request("GET", url, headers = {
  "auth-token": ELETRICITY_API
})

print(response.text)
```
